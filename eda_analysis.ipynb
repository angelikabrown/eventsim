```python
import pandas as pd
```

```python
df = pd.read_json('samples/2023061719_20.json', lines=True)
df.head(5)
```

```python
df.describe()
```

```python

# Count free vs paid in the 'level' column
level_counts = df['level'].value_counts()
print(level_counts)
```

```python
#unique values in userAgent column
user_agent_counts = df['userAgent'].value_counts()
print(user_agent_counts)
```

```python
# Count occurrences of each userId and display the top 10
top_user_ids = df['userId'].value_counts().head(10)
print(top_user_ids)
```

```python
top_user_df = df[df['userId'].isin(top_user_ids)]
top_user_df.head(10)
```

```python
# Print unique values for each column in top_user_df
for column in top_user_df.columns:
    print(f"Unique values in column '{column}':")
    print(top_user_df[column].unique())
    print()
```

```python
# Print unique userAgent values in top_user_df
print("Unique userAgent in top_user_df:")
print(top_user_df['userAgent'].unique())
```

```python
# Print unique state values in top_user_df
print("Unique states in top_user_df:")
print(top_user_df['state'].unique())
```

```python
# Count how many of the top 10 users are 'paid' and 'free'
top_user_level_counts = top_user_df['level'].value_counts()
print("Counts of 'paid' and 'free' for top 10 users:")
print(top_user_level_counts)
```

```python
# Count how many of the top 10 users are male versus female
top_user_gender_counts = top_user_df['gender'].value_counts()
print("Counts of male and female for top 10 users:")
print(top_user_gender_counts)
```
